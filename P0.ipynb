## Exercise 3: separación balanceada de conjuntos de entrenamiento y test


*   Partir de los datos de Iris (empleados en el Exercise 1), y separar en training (80\% de los datos) y test (20\%) aleatoriamente, conservando la proporción de elementos en cada clase tanto en training como en test. Con esto se pretende evitar que haya clases infra-representadas en entrenamiento o test. Con "aleatoriamente" nos referimos a que no sería válido escoger como conjunto de entrenamiento el 80\% inicial de ejemplos de cada clase y como conjunto de test el 20\% final. Es decir, al principio, los ejemplos pertenecientes a cada clase deben desordenarse con respecto al orden original. En la implementación de este exercise no se pueden emplear funciones como $train\_test\_split$ de scikit-learn o similares.
*   Se debe imprimir el número resultante de ejemplos de cada clase, tanto en entrenamiento como en test, así como las clases de cada uno de los ejemplos de entrenamiento y test. En la celda de resultados que se incluye a continuación se muestra el tipo de salida que se espera obtener.





Para este exercise se crean indices aleatorios entre 0 y el numero de items de 'x_iris'. Que son desordenados para así cumplir con la aleatoriedad exigida.


```python
np.random.seed(0)

indices = np.arange(0, len(x_iris))
np.random.shuffle(indices)
```

Se crean 2 vectores auxiliares (x_aux e y_aux) a partir de los originales pero utilizando el orden de los indices antes obtenidos.


```python
x_aux = x_iris[indices]
y_aux = y_iris[indices]
```

El tamaño de los datos del training de cada clase se determina dividiendo el total de los datos entre el numero de clases existentes y multiplicandose por el porcentaje pedido, en este caso 80%.


```python
tam_train = int(len(x_iris) / 3 * 0.8)
```

Para obtener la lista de training se crea un bucle que recorre los indices aleatorios y guarda el indice si no se han guardado los suficientes de su clase.


```python
n_seto, n_ver, n_vir = 0, 0, 0
train_list, test_list, ind_list = [], [], []

for i in indices:
    if y_aux[i] == 0 and n_seto < tam_train: # si es de una clase y no hemos llegado al limite la aceptamos
      train_list.append(i)
      n_seto += 1
      ind_list.append(i) # añadimos la posicion a una lista auxiliar
    elif y_aux[i] == 1 and n_ver < tam_train:
      train_list.append(i)
      n_ver += 1
      ind_list.append(i)
    elif y_aux[i] == 2 and n_vir < tam_train:
      train_list.append(i)
      n_vir += 1
      ind_list.append(i)
```

Añadimos las posiciones restantes a la lista de test borrando las ya visitadas (80%)


```python
test_list = np.delete(indices, ind_list)
```

Finalmente se obtiene el output:


```python
print("--- Clase setosa ---")
print("Ejemplos train: {}".format(n_seto))
print("Ejemplos test: {}".format(n_seto - len(test_list)))
print("--- Clase versicolor ---")
print("Ejemplos train: {}".format(n_ver))
print("Ejemplos test: {}".format(n_ver - len(test_list)))
print("--- Clase virginica ---")
print("Ejemplos train: {}".format(n_vir))
print("Ejemplos test: {}".format(n_vir - len(test_list)))
print("Clase de los ejemplos de entrenamiento:\n {}".format(y_iris[train_list]))
print("Clase de los ejemplos de test:\n {}".format(y_iris[test_list]))
```

Funciones utilizadas:


```python
def f1(x, y):
    return 10 - (2 * abs(x + y)) + (2 * abs(y - x))

def f2(x,y):
    return 10 * np.sin(np.sqrt(x**2 + y**2))
```

Creación de la grafica:


```python
import matplotlib.cm as cm

#Se crea la figura que albergará la gráfica
fig = plt.figure(figsize=(30, 13))

#Función 1

#Los valores tendrán un rango de valores entre -6 y 6
#Serán 30 valores en ese rango.
a = np.linspace(-6, 6, 30)
b = np.linspace(-6, 6, 30)

#Se consiguen las coordenadas 3D
a, b = np.meshgrid(a, b)

#Se representa la función en la primera posición
ax = fig.add_subplot(1, 2, 1, projection='3d')

#Se añade el titulo utilizando Latex 
ax.set_title(r'$f_{1}(x,y) = 10 - 2 · \left | {x + y} \right | + 2 · \left | {y - x} \right |$',  fontname="Times New Roman", fontsize=25, fontweight="bold")

#Finalmente se dibuja esta primera función
ax.plot_surface(a, b, f1(a,b), cmap=cm.coolwarm)



# Función 2

#Los valores tendrán un rango de valores entre -2 y 2
#Serán 100 valores en ese rango.
c = np.linspace(-2, 2, 100)
d = np.linspace(-2, 2, 100)

#Se consiguen las coordenadas 3D
c, d = np.meshgrid(c, d)


#Se representa la función en la segunda posición
ax = fig.add_subplot(1, 2, 2, projection='3d')

#Se añade el titulo utilizando Latex 
ax.set_title(r'$f_{2}(x,y) = 10 · sin \left ( \sqrt {x^2 + y^2} \right ) $',fontname="Times New Roman", fontsize=25, fontweight="bold")

#Finalmente se dibuja esta segunda función
ax.plot_surface(c, d, f2(c,d), cstride=1, rstride=1, cmap=cm.viridis)

#Muestra el resultado final
plt.show()
     
#NOTA: puede que Colab proporcione el siguiente error al intentar poner el título de la figura en LaTeX:
#[Errno 2] No such file or directory: 'latex': 'latex'.
#Una posible solución sería la indicada en el siguiente enlace: 
#https://stackoverflow.com/questions/55746749/latex-equations-do-not-render-in-google-colaboratory-when-using-matplotlib
#Y, más concretamente, serviría con hacer
#!apt install texlive-fonts-recommended texlive-fonts-extra cm-super dvipng
```
